```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(4)
qc.qubits
```

A quantum register is the quantum equivalent of a processor register in classical computing. It is a collection or ordered sequence of multiple qubits grouped together to hold quantum information, serving as the fundamental working memory where quantum algorithms and operations are
While a classical register can only store a single definitive sequence of 0s and 1s, a quantum register uses the principles of superposition and entanglement to do much more

```python
from qiskit.circuit import QuantumRegister,ClassicalRegister
qc1=QuantumRegister(2, "qr1")
cr1=ClassicalRegister(2, 'cr1')
qc2=QuantumRegister(4, 'qr2')
```

```python
combined_cir= QuantumCircuit(qc1, qc2,cr1)
combined_cir.qubits
```

```python
#checking,,,index and register by using the circuit's find_bit method and its attributes.
find_qc= qc2[2]
print('index', combined_cir.find_bit(find_qc).index)
```

```python
find_cr= cr1[1]
print('index', combined_cir.find_bit(find_cr).index)
```
