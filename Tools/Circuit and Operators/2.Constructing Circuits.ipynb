```python
# Importing library
from qiskit import QuantumCircuit, QuantumRegister
```

```python
qc = QuantumCircuit(3)
qc.qubits
```

register=(2, "q"): This indicates the register consists of 2 qubits and is labeled with the string name "q".

index=0 and index=1: These are the specific positions of the qubits within that register. In quantum circuits, we almost always use zero-based indexing.

```python
# Indcating qubit and its registry location
qubit_0 = qc.qubits[0] # <Qubit register=(2, "q"), index=0>
qubit_1 = qc.qubits[1]
qubit_1 = qc.qubits[2]# <Qubit register=(2, "q"), index=1>
qc.draw('mpl')
```

#### Multiple QuantumRegister and ClassicalRegister objects can be combined to create a circuit. 
Every QuantumRegister and ClassicalRegister can also be named.

```python
# A quantum register is a collection of qubits (quantum bits) used to store and process quantum information
from qiskit.circuit import QuantumRegister
qr1= QuantumRegister(4, "qregister1") #quantum bit
cr1= QuantumRegister(1, "qregister2") #classical bit
combined= QuantumCircuit(qr1,cr1)
combined.qubits
```

```python
combined.draw()
```

```python
# Circuit with 2 QuantumRegisters and 1 ClassicalRegister
qr1 = QuantumRegister(4, "qreg1")  # Create a QuantumRegister with 2 qubits
#qr2 = QuantumRegister(1, "qreg2")  # Create a QuantumRegister with 1 qubit
#cr1 = ClassicalRegister(3, "creg1")  # Create a ClassicalRegister with 3 cbits
 
combined_c = QuantumCircuit(qr1, cr1)
combined_c.qubits
```

```python
combined_c.draw()
```
