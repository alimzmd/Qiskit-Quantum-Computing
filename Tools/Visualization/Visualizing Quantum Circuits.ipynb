```python
import qiskit
from qiskit import QuantumCircuit
```

```python
circuit = QuantumCircuit(3, 3)
circuit.x(1)
circuit.h(range(3))
circuit.cx(0, 1)
circuit.measure(range(3), range(3));
```

```python
# with print method, --effect
print(circuit)
```

```python
circuit.draw()  #no effect
```

```python
# Matplotlib drawing
circuit.draw(output="mpl")
```
